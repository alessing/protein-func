```python
import matplotlib.pyplot as plt

# Data for the models
models = ["full_rank", "blks_4", "bases_1", "lora"]
test_acc = [0.985, 0.98, 0.975, 0.985]
num_params = [288897, 141441, 157837, 163969]

# Define light colors for each model
scatter_colors = ['lightblue', 'lightcoral', 'lightgreen', 'lightskyblue']

# Scatter Plot for Test Accuracy vs Number of Parameters with different colors for each model
plt.figure(figsize=(8, 6))
for i, model in enumerate(models):
    plt.scatter(num_params[i], test_acc[i], color=scatter_colors[i], label=model)

# Annotate each point
for i, model in enumerate(models):
    plt.annotate(model, (num_params[i], test_acc[i]), textcoords="offset points", xytext=(0, 5), ha='center')

# Set labels and title
plt.xlabel("Number of Parameters")
plt.ylabel("Test Accuracy")
plt.title("Test Accuracy vs Number of Parameters")
plt.legend()

# Show plot
plt.grid(True)
plt.show()

# Performance Efficiency Calculation (Test Accuracy / Number of Parameters)
performance_efficiency = [acc / params for acc, params in zip(test_acc, num_params)]

# Define light colors for the bar plot
bar_colors = ['lightblue', 'lightcoral', 'lightgreen', 'lightskyblue']

# Bar Plot for Performance Efficiency with different colors for each model
plt.figure(figsize=(8, 6))
plt.bar(models, performance_efficiency, color=bar_colors)

# Set labels and title
plt.xlabel("Models")
plt.ylabel("Performance Efficiency (Test Accuracy / Number of Parameters)")
plt.title("Performance Efficiency of Different Models")

# Show plot
plt.grid(True)
plt.show()
```

```python
# Calculating Performance Efficiency for each model
performance_efficiency = [model["best_test_acc"] / model["num_params"] for model in models.values()]

# Creating a bar plot for Performance Efficiency
plt.figure(figsize=(10, 6))
plt.bar(model_names, performance_efficiency, color=colors, alpha=0.7)

# Labeling axes and title
plt.xlabel("Models", fontsize=14)
plt.ylabel("Performance Efficiency (Accuracy / Parameters)", fontsize=14)
plt.title("Performance Efficiency of Models", fontsize=16)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data for the models with means and standard deviations
models = {
    "full_rank": {
        "best_test_acc_mean": 0.8105263157894737,
        "best_test_acc_std": 0.05367388961676616,
        "num_params": 516097,
    },
    "blks_4": {
        "best_test_acc_mean": 0.8631578947368421,
        "best_test_acc_std": 0.042105263157894736,
        "num_params": 221185,
    },
    "bases_1": {
        "best_test_acc_mean": 0.8421052631578947,
        "best_test_acc_std": 0.07443229275647868,
        "num_params": 221209,
    },
    "lora": {
        "best_test_acc_mean": 0.8421052631578947,
        "best_test_acc_std": 0.0332871332649303,
        "num_params": 245761,
    },
}

# Extracting data for plotting
num_params = [model["num_params"] for model in models.values()]
test_acc_means = [model["best_test_acc_mean"] for model in models.values()]
test_acc_stds = [model["best_test_acc_std"] for model in models.values()]
model_names = list(models.keys())

# Assign unique colors to each model
colors = ['red', 'green', 'blue', 'orange']

# Creating the scatter plot with error bars
plt.figure(figsize=(10, 6))
for i, name in enumerate(model_names):
    plt.errorbar(
        num_params[i],
        test_acc_means[i],
        yerr=test_acc_stds[i],
        fmt="o",
        label=name,
        color=colors[i],
        capsize=5,
    )

# Adding legend
plt.legend(title="Models", fontsize=12)

# Labeling axes and title
plt.xlabel("Number of Parameters", fontsize=14)
plt.ylabel("Test Accuracy", fontsize=14)
plt.title("Test Accuracy vs Number of Parameters with Std Dev", fontsize=16)
plt.grid(alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data for the models with means and standard deviations
models = {
    "full_rank": {
        "best_test_acc_mean": 0.8105263157894737,
        "best_test_acc_std": 0.05367388961676616,
        "num_params": 516097,
    },
    "blks_4": {
        "best_test_acc_mean": 0.8631578947368421,
        "best_test_acc_std": 0.042105263157894736,
        "num_params": 221185,
    },
    "bases_1": {
        "best_test_acc_mean": 0.8421052631578947,
        "best_test_acc_std": 0.07443229275647868,
        "num_params": 221209,
    },
    "lora": {
        "best_test_acc_mean": 0.8421052631578947,
        "best_test_acc_std": 0.0332871332649303,
        "num_params": 245761,
    },
}

# Calculating performance efficiency and its standard deviation
efficiency_means = [
    model["best_test_acc_mean"] / model["num_params"] for model in models.values()
]
efficiency_stds = [
    model["best_test_acc_std"] / model["num_params"] for model in models.values()
]
model_names = list(models.keys())

# Assign unique colors to each model
colors = ['red', 'green', 'blue', 'orange']

# Creating a bar plot with error bars for performance efficiency
plt.figure(figsize=(10, 6))
plt.bar(
    model_names,
    efficiency_means,
    yerr=efficiency_stds,
    color=colors,
    alpha=0.7,
    capsize=5,
)

# Labeling axes and title
plt.xlabel("Models", fontsize=14)
plt.ylabel("Performance Efficiency (Mean Accuracy / Parameters)", fontsize=14)
plt.title("Performance Efficiency of Models with Std Dev", fontsize=16)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
